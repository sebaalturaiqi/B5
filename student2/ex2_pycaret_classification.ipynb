## Palmer Penguins Dataset

The goal of palmer penguins is to provide a great dataset for data exploration & visualization, as an alternative to iris.

The data contains 344 penguins. There are 3 different species of penguins in this dataset, collected from 3 islands in the Palmer Archipelago, Antarctica.

- Features: `4` numerical, `2` categorical
- Target: `species` (Categorical / 3 classes)
- Size: `344` samples
- Source: [Palmer Penguins](https://allisonhorst.github.io/palmerpenguins/)

### Load the data

```python
import seaborn as sns

penguins = sns.load_dataset('penguins')
penguins
```

```python

from sklearn.model_selection import train_test_split
from autogluon.tabular import TabularPredictor
data_exp1 = penguins[['flipper_length_mm', 'bill_length_mm', 'species']]
train_1, test_1 = train_test_split(data_exp1, test_size=0.2, random_state=42)
pred_1 = TabularPredictor(label='species').fit(train_1, time_limit=60, presets='medium_quality')
pred_1.leaderboard(test_1)
```

## Experiments

1. **First experiment**:
    - X = Flipper Length (numerical) & Bill Length (numerical)
    - y = Species (categorical)
2. **Second experiment**:
    - X = Weights (numerical) & Species (categorical)
    - y = Sex (categroical)
3. **Third experiment**:
    - X = `island`, `bill_length_mm`, `bill_depth_mm`, `flipper_length_mm`, `body_mass_g`
    - y = Sex and Species (multi-label classification)

```python

data_exp1 = penguins[['flipper_length_mm', 'bill_length_mm', 'species']]
train_1, test_1 = train_test_split(data_exp1, test_size=0.2, random_state=42)
pred_1 = TabularPredictor(label='species').fit(train_1, time_limit=60, presets='medium_quality')
pred_1.leaderboard(test_1)
```

```python
data_exp2 = penguins[['body_mass_g', 'species', 'sex']]
data_exp2 = data_exp2.dropna(subset=['sex'])
from sklearn.model_selection import train_test_split
train_2, test_2 = train_test_split(data_exp2, test_size=0.2, random_state=42)
from autogluon.tabular import TabularPredictor
pred_2 = TabularPredictor(label='sex').fit(train_2, time_limit=60, presets='medium_quality')
pred_2.leaderboard(test_2)
```

```python

features = ['island', 'bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']

print("TYPE OF FEATURES: ", type(features))
data_3a = penguins[features + ['species']].dropna()
train_3a, test_3a = train_test_split(data_3a, test_size=0.2, random_state=42)
pred_3a = TabularPredictor(label='species').fit(train_3a, time_limit=30, presets='medium_quality')
pred_3a.leaderboard(test_3a)

print("\n GENDER")
data_3b = penguins[features + ['sex']].dropna()
train_3b, test_3b = train_test_split(data_3b, test_size=0.2, random_state=42)
pred_3b = TabularPredictor(label='sex').fit(train_3b, time_limit=30, presets='medium_quality')
pred_3b.leaderboard(test_3b)
```
