### Step 4. How many products cost more than $10.00?

```python
result = chipo[chipo['item_price'] > 10]
print(result.head())
```

### Step 5. What is the price of each item? 
###### print a data frame with only three columns item_name choice_description and product_price

```python
chipo['product_price'] = chipo['item_price']

result = chipo[['item_name', 'choice_description', 'product_price']]

print(result.head())
```

### Step 6. Sort by the name of the item

```python
sorted_items = chipo.sort_values(by='item_name')

print(sorted_items.head())
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
most_expensive = chipo.sort_values(by='item_price', ascending=False)

print("أغلى منتج:")
print(most_expensive.head(1))

print("الكمية:", most_expensive.iloc[0]['quantity'])
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie = chipo[chipo['item_name'] == 'Veggie Salad Bowl']

print("عدد الطلبات:", len(veggie))
print(veggie.head())
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
soda = chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)]

print("عدد الطلبات:", len(soda))
print(soda.head())
```
