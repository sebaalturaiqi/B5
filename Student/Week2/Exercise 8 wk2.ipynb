### Step 1. Import the necessary libraries

```python
import pandas as pd
print("Libraries imported successfully")
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user"
users = pd.read_csv(url, sep='|')
print("Dataset loaded successfully:\n")
print(users.head())  
```

### Step 4. Discover what is the mean age per occupation

```python
mean_age = users.groupby('occupation')['age'].mean()

print("Mean age per occupation:\n")
print(mean_age)
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
male_ratio = users[users['gender'] == 'M'].groupby('occupation')['gender'].count() / users.groupby('occupation')['gender'].count()

male_ratio_sorted = male_ratio.sort_values(ascending=False)

print("Male ratio per occupation (sorted):\n")
print(male_ratio_sorted)
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
min_max_age = users.groupby('occupation')['age'].agg(['min', 'max'])

print("Min and Max age per occupation:\n")
print(min_max_age)
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
mean_age_gender = users.groupby(['occupation', 'gender'])['age'].mean()

print("Mean age per occupation and gender:\n")
print(mean_age_gender)
```

### Step 8.  For each occupation present the percentage of women and men

```python
gender_count = users.groupby(['occupation', 'gender']).size()

total_per_occupation = users.groupby('occupation').size()

gender_percentage = gender_count / total_per_occupation

print("Percentage of men and women per occupation:\n")
print(gender_percentage)
```
