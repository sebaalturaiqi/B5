# Exam scores

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(df["Exam_Score"], bins=20)
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Count")
plt.show()
```

### Insight

The distribution of exam scores appears to be approximately normal 

Most students scored between 60 and 75, with the peak around 65–70.

There are very few students with extremely high or very low scores, which indicates that the data is fairly balanced without extreme outliers.

This suggests that the majority of students perform within an average range.

تحليل العلاقه 
# hours stdied vs exam score

```python
sns.regplot(x="Hours_Studied", y="Exam_Score", data=df)
plt.title("Hours Studied vs Exam Score")
plt.show()
```

### Observation

There is a positive relationship between study hours and exam score.

Students who study more tend to get higher scores.

There are some variations, but overall the trend supports our hypothesis.

# sleeping hours vs exam score

```python
sns.regplot(x="Sleep_Hours", y="Exam_Score", data=df)
plt.title("Sleep Hours vs Exam Score")
plt.show()
```

### Observation

The relationship between sleep hours and exam score appears to be weak.

The trend line is nearly flat, indicating that sleep hours do not have a strong direct impact on exam performance.

Students with both low and high sleep hours show similar score distributions.

This suggests that sleep alone is not a strong predictor of exam scores, and other factors may have a greater influence.

## Attendance vs Exam Score

We analyze whether attendance affects exam performance.

```python
df["Attendance_Group"] = pd.cut(df["Attendance"], bins=5)

df.groupby("Attendance_Group")["Exam_Score"].mean().plot(kind="bar")

plt.title("Average Exam Score by Attendance Group")
plt.xlabel("Attendance Groups")
plt.ylabel("Average Exam Score")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Observation

The bar chart shows a clear positive relationship between attendance and exam score.

As attendance increases, the average exam score also increases.

Students in higher attendance groups tend to achieve better results compared to those with lower attendance.

This indicates that attendance is an important factor influencing student performance.

# متوسط الدرجات (Low,Meduim, High)

```python
plt.figure(figsize=(6,4))

sns.barplot(x="Motivation_Level", y="Exam_Score", data=df)

plt.ylim(60, 75)  
plt.title("Average Exam Score by Motivation Level")
plt.xlabel("Motivation Level")
plt.ylabel("Average Exam Score")

plt.show()
```

### Observation

There is a slight increase in exam scores as motivation level increases.

However, the difference between groups is small, suggesting that motivation has a limited impact compared to other factors.

## Correlation Heatmap

We analyze the relationships between all numerical variables.

```python
corr = df.corr(numeric_only=True)["Exam_Score"].sort_values()

corr.plot(kind="bar")
plt.title("Correlation with Exam Score")
plt.ylabel("Correlation")
plt.show()
```

### Observation

- Hours_Studied shows a positive correlation with Exam_Score.
- Attendance also has a positive relationship with Exam_Score.
- Some variables have weak or no correlation.
