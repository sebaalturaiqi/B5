```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
plt.style.use('seaborn-v0_8-darkgrid')
np.random.seed(42)
```

```python
df = pd.read_csv('WA_Marketing-Campaign.csv')
```

```python
df.shape
df.head()
df.info()
df.describe()
df.isnull().sum()
```

# checking the data
- The dataset contains information about promotions, stores, and sales.
- We checked the structure and data types.
- No obvious loading issues detected.

```python
sns.countplot(x='Promotion', data=df)
plt.title("Distribution of Promotions")
plt.xlabel("Promotion Type")
plt.ylabel("Count")
plt.show()
```

### Insight

The distribution shows how many observations belong to each promotion.

This helps us understand if the data is balanced across the three promotions.

 # توزيع المبيعات (Sales Distribution)

```python
sns.histplot(df['SalesInThousands'], kde=True)
plt.title("Sales Distribution")
plt.xlabel("Sales (Thousands)")
plt.ylabel("Frequency")
plt.show()
```

### Sales Distribution Insight

The sales distribution appears slightly right-skewed.

Most sales values fall between 40 and 60, while there are some higher values extending toward 100.

This suggests that while most locations perform moderately, a few locations achieve significantly higher sales.

# boxplot

```python
sns.boxplot(x=df['SalesInThousands'])
plt.title("Boxplot of Sales")
plt.show()
```

### Boxplot Insight

The boxplot shows that most sales values are concentrated in a central range.

However, there are several high-value outliers, indicating that some locations perform exceptionally well compared to others.

These outliers may represent highly successful stores or favorable market conditions.

# Bivariate Analysis

Promotion vs Sales

```python
sns.boxplot(x='Promotion', y='SalesInThousands', data=df)
plt.title("Sales by Promotion")
plt.xlabel("Promotion")
plt.ylabel("Sales (Thousands)")
plt.show()
```

### Sales by Promotion Insight

The boxplot shows clear differences in sales performance across the three promotions.

- Promotion 1 has the highest median sales, indicating the best overall performance
- Promotion 3 performs moderately, close to Promotion 1
- Promotion 2 has the lowest median sales, suggesting weaker performance

Additionally, all promotions show some high-value outliers, but Promotion 1 consistently achieves higher typical sales.

This suggests that Promotion 1 is the most effective campaign.

# Average Sales Insight

```python
df.groupby('Promotion')['SalesInThousands'].mean()
```

### Average Sales Insight

The average sales clearly show differences between the three promotions.

- Promotion 1 achieved the highest average sales (~58.7), making it the best performing campaign
- Promotion 3 follows with slightly lower performance (~55.4)
- Promotion 2 has the lowest average sales (~47.3), indicating the weakest performance

These results confirm that Promotion 1 is the most effective strategy for increasing sales.

# analysis based on time 

```python
sns.lineplot(x='week', y='SalesInThousands', hue='Promotion', data=df)
plt.title("Sales Trend by Week")
plt.xlabel("Week")
plt.ylabel("Sales (Thousands)")
plt.show()
```

### Sales Trend Insight

The line plot shows how each promotion performs over the four weeks.

We can observe whether performance is consistent or changes over time.

Consistent performance indicates a reliable marketing strategy.

# Summary & Conclusion

In this analysis, we evaluated the performance of three different marketing promotions using sales data across multiple locations and weeks.

## Key Findings

- Promotion 1 achieved the highest average sales (~58.7), making it the top-performing campaign
- Promotion 3 performed moderately well, ranking second (~55.4)
- Promotion 2 had the lowest performance (~47.3), indicating it is the least effective

- Sales distribution showed some high-performing locations (outliers), but overall trends remained consistent
- The weekly trend analysis showed that Promotion 1 maintained strong performance over time

## Final Recommendation

Based on the analysis, Promotion 1 is the most effective marketing strategy and should be selected to maximize sales.
