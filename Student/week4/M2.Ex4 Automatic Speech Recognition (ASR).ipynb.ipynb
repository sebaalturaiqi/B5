```python
import os
from langchain_community.document_loaders import YoutubeLoader
from docling.datamodel.base_models import InputFormat
from docling.document_converter import DocumentConverter

def solve_task(video_url):
    """
    Implements a robust YouTube ASR system:
    1. Attempts to fetch existing transcripts via LangChain.
    2. Falls back to Docling's ASR pipeline if no transcript is found.
    """

    # Step 1: Attempt to fetch existing transcripts using LangChain
    try:
        print(f"🔍 Searching for existing transcript for: {video_url}")

        loader = YoutubeLoader.from_youtube_url(video_url, add_video_info=True, language=["en", "ar"])
        docs = loader.load()
        if docs:
            print("✅ Transcript found successfully!")
            return docs[0].page_content
    except Exception as e:
        print(f"⚠️ No transcript available or error: {e}")

    # Step 2: Fallback to Docling ASR Pipeline
    print("🎙️ Starting Docling ASR Pipeline (Audio-to-Text)...")

    # Download audio using yt-dlp
    audio_filename = "audio_temp.mp3"

    get_ipython().system(f'yt-dlp -x --audio-format mp3 -o {audio_filename} {video_url}')

    # Initialize Docling Converter for transcription
    converter = DocumentConverter()
    result = converter.convert(audio_filename)

    print("✅ Speech recognition completed successfully!")
    return result.document.export_to_markdown()

# --- Execution ---

video_link = "https://www.youtube.com/watch?v=PeMlggyqz0Y"
final_result = solve_task(video_link)

print("\n" + "="*30)
print("FINAL OUTPUT (Transcription)")
print("="*30 + "\n")
print(final_result)
```
